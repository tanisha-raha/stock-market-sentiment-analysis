# Market analysis
Daily posts are mapped to the next trading day. This analysis needs internet for yfinance and `date`/`ticker` in the CSV.

```python
from src.data_loader import load_dataset, require_market_columns
from src.sentiment_index import daily_sentiment_index
from src.market_data import fetch_market_data, align_sentiment_to_trading_days
from src.signals import lag_analysis
df, _ = load_dataset('../data/raw/dataset.csv'); require_market_columns(df)
daily = daily_sentiment_index(df[df.ticker == 'NVDA'])
market = fetch_market_data('NVDA', daily.date.min(), daily.date.max() + __import__('pandas').Timedelta(days=7))
merged = align_sentiment_to_trading_days(daily, market).merge(market, on='date')
lag_analysis(merged)
```
